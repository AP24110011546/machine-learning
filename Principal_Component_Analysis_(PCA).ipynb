```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}

df = pd.DataFrame(data)

# Step 1: Standardize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Step 2: Apply PCA
pca = PCA(n_components=2)
pca_data = pca.fit_transform(scaled_data)

# Step 3: Display results
print("Original Data:")
print(df)

print("\nPCA Transformed Data:")
print(pca_data)

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_)

print("\nTotal Variance Preserved:")
print(sum(pca.explained_variance_ratio_))
```

This Python program uses Scikit-learn to perform Principal Component Analysis (PCA) for dimensionality reduction. It standardizes the data, transforms the original features into principal components, and calculates the explained variance ratio to determine how much information is preserved.
