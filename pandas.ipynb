Column Selection

```python
import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),
'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print ("Add new column by passing series")
df['three']=pd.Series([20,40,60],index=['a','b','c'])
print (df)
print ("Add new column using existing DataFrame columns")
df['four']=df['one']+df['three']
print (df)
```

Column Addition

```python
import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print ("Add new column by passing series")
df['three']=pd.Series([20,40,60],index=['a','b','c'])
print (df)
print ("Add new column using existing DataFrame columns")
df['four']=df['one']+df['three']
print (df)
```

a DataFrame is made with columns one and two. Then a new column three is added using a Series that only has values for a, b, c, so the rest show NaN. Lastly, column four is created by adding one and three together, and where either has NaN, the result is also NaN.

Column Deletion

```python
import pandas as pd
info = {'one' : pd.Series([1, 2], index= ['a', 'b']),'two' : pd.Series([1, 2, 3], index=['a', 'b', 'c'])}
df = pd.DataFrame(info)
print ("The DataFrame:")
print (df)
print ("Delete the first column:")
del df['one']
print (df)
print ("Delete the another column:")
df.pop('two')
print (df)
```

del df['one'] is used to remove the one column and df.pop('two') is used to delete the two column which leaves an empty dataframe at the end

Selection by Label

loc function

```python
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print("Data frame:")
print(df)
print("b index elements are:")
print (df.loc['b'])
print("e index elements are:")
print(df.loc['e'])
```

when a dataframe is created with columns one and two then by using df.loc['b '] and df.loc['e'] , the values that are present in the b and e are shown row wise

iloc functions

```python
import pandas as pd
info = {'one' : pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e']),'two' : pd.Series([1, 2, 3, 4, 5, 6], index=['a', 'b', 'c', 'd', 'e', 'f'])}
df = pd.DataFrame(info)
print(df)
print (df.iloc[3])
```

by using df.iloc it picks out the row at position 3 so it shows the values at that particular row position

Deletion of rows

```python
import pandas as pd
a_info = pd.DataFrame([[4, 5], [6, 7]], columns = ['x','y'])
b_info = pd.DataFrame([[8, 9], [10, 11]], columns = ['x','y'])
a_info = pd.concat([a_info,b_info])
print(a_info)
a_info = a_info.drop(0)
print('Drop the row with label 0')
print(a_info)
```

two dataframes with columns x and y are joined together using pd.concat(). then a_info.drop(0) removes all rows labeled 0

DataFrame.drop_duplicates

```python
import pandas as pd
emp = {"Name": ["Siva", "Rama", "Krishna", "Siva"],
"Age": [21, 32, 29, 21]}
info = pd.DataFrame(emp)
print(info)
info = info.drop_duplicates()
print("After removing the duplicates the dataframe is :")
print(info)
```

Null values identification in given dataset


```python
import pandas as pd
import numpy as np
info = pd.DataFrame({"Person":["Siva", "Pavan", "Gopal", "Raghu"],
"Age": [27., 29, np.nan, 32]})
print(info)
print(info.count())
print("null values")
print(info.isnull())
print(info.isnull().sum())
```

Sorting order




```python
import pandas as pd
import numpy as np
info = pd.DataFrame({"Person":["Siva", "Pavan", "Gopal", "Raghu"],
"Age": [27., 29, np.nan, 32]})
print(info)
print("Sorting Order")
print(info.sort_index())
print("Descending Order")
print(info.sort_index(ascending=False))
```

Aggregation functions

```python
import pandas as pd
df=pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")
print(df)
mean_df = df.groupby('species').mean()
print("Mean values for each species:")
print(mean_df)
sum_df = df.groupby('species').sum()
print("\nSum of values for each species:")
print(sum_df)
```
