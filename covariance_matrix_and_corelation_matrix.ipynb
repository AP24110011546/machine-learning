```python
import pandas as pd

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}

df = pd.DataFrame(data)

# Compute covariance matrix
print("Covariance Matrix:")
print(df.cov())

# Compute correlation matrix
print("\nCorrelation Matrix:")
print(df.corr())
```

This Python program uses Pandas to compute the covariance and correlation matrices for multiple attributes in a dataset. The covariance matrix shows how attributes vary together, while the correlation matrix shows the strength and direction of their linear relationships.
