```python
import pandas as pd

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}

df = pd.DataFrame(data)

# Compute covariance matrix
print("Covariance Matrix:")
print(df.cov())

# Compute correlation matrix
print("\nCorrelation Matrix:")
print(df.corr())
```

This Python program uses Pandas to calculate the covariance and Pearson correlation coefficient between two attributes. Covariance indicates how two variables vary together, while correlation measures the strength and direction of their linear relationship.
